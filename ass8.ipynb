Write a unique paragraph (5–6 sentences) about your favourite topic.
1. Convert text to lowercase and remove punctuation using re.
2. Tokenize the text into words and sentences.
3. Split using split() and word_tokenize() and compare how Python split and NLTK's word_tokenize() differ.
4. Remove stopwords using NLTK's stopwords list.
5. Display word frequency distribution (excluding stopwords).

```python

import nltk
import re

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

# Downloading required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python

# Writing a paragraph about Artificial Intelligence

text = """Artificial Intelligence is transforming the modern world.
It helps people solve complex problems and automate daily tasks.
AI is widely used in healthcare, education, business, and technology.
Machine learning allows computers to learn from data and make predictions.
I believe artificial intelligence will create many opportunities in the future.
It makes technology smarter, faster, and more useful."""

print("Original Paragraph:\n", text)

# 1. Convert to lowercase and remove punctuation
clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("\nCleaned Text:\n", clean_text)

# 2. Tokenizing into sentences and words
sentences = sent_tokenize(text)
words = word_tokenize(clean_text)

print("\nSentences:")
print(sentences)

print("\nWord Tokens:")
print(words)

# 3. Comparing split() and word_tokenize()
split_words = text.split()
nltk_words = word_tokenize(text)

print("\nUsing split():")
print(split_words)

print("\nUsing word_tokenize():")
print(nltk_words)

# 4. Removing stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in words
    if word not in stop_words
]

print("\nWords after removing stopwords:")
print(filtered_words)

# 5. Word frequency distribution
frequency = FreqDist(filtered_words)

print("\nWord Frequency Distribution:")
print(frequency.most_common())
```

Using the same paragraph from Q1:
1. Extract all words with only alphabets using re.findall().
2. Remove stopwords using NLTK's stopword list.
3. Perform stemming with PorterStemmer.
4. Perform lemmatization with WordNetLemmatizer.
5. Compare the stemmed and lemmatized outputs and explain when you would prefer one over the other.

```python

from nltk.stem import PorterStemmer, WordNetLemmatizer

# Downloading WordNet resources
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger_eng')

# 1. Extracting words containing only alphabets
alphabet_words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Alphabetic Words:")
print(alphabet_words)

# 2. Removing stopwords
stop_words = set(stopwords.words('english'))

filtered_words = [
    word for word in alphabet_words
    if word not in stop_words
]

print("\nWords after Stopword Removal:")
print(filtered_words)

# 3. Porter Stemming
porter = PorterStemmer()

stemmed_words = [
    porter.stem(word) for word in filtered_words
]

print("\nStemmed Words:")
print(stemmed_words)

# 4. Lemmatization
lemmatizer = WordNetLemmatizer()

lemmatized_words = [
    lemmatizer.lemmatize(word)
    for word in filtered_words
]

print("\nLemmatized Words:")
print(lemmatized_words)

# 5. Comparing both outputs
comparison = []

for original, stem, lemma in zip(
    filtered_words, stemmed_words, lemmatized_words
):
    comparison.append({
        "Original Word": original,
        "Stemmed Word": stem,
        "Lemmatized Word": lemma
    })

import pandas as pd

comparison_df = pd.DataFrame(comparison)

print("\nComparison Table:")
display(comparison_df)
```

Choose 3 short texts of your own (e.g., different news headlines, product reviews).
1. Use CountVectorizer to generate the Bag of Words representation.
2. Use TfidfVectorizer to compute TF-IDF scores.
3. Print and interpret the top 3 keywords from each text using TF-IDF.

```python

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd

# Three short texts about technology

documents = [
    "Artificial intelligence helps computers learn and solve problems.",
    "Machine learning improves artificial intelligence through data.",
    "Technology and computers are changing the modern world."
]

# 1. Bag of Words using CountVectorizer
count_vectorizer = CountVectorizer()

bow_matrix = count_vectorizer.fit_transform(documents)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Matrix:")
print(bow_matrix.toarray())

bow_df = pd.DataFrame(
    bow_matrix.toarray(),
    columns=count_vectorizer.get_feature_names_out()
)

display(bow_df)

# 2. TF-IDF calculation
tfidf_vectorizer = TfidfVectorizer()

tfidf_matrix = tfidf_vectorizer.fit_transform(documents)

print("\nTF-IDF Vocabulary:")
print(tfidf_vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())

# 3. Finding top 3 keywords from each text
feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTOP 3 KEYWORDS FROM EACH TEXT")

for i, row in enumerate(tfidf_matrix.toarray()):

    top_indices = row.argsort()[-3:][::-1]

    print("\nText", i + 1)

    for index in top_indices:
        print(feature_names[index], ":", round(row[index], 4))
```

Write two short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).
1. Preprocess and tokenize both texts.
2. Calculate:
a. Jaccard Similarity using sets.

b. Cosine Similarity using TfidfVectorizer and cosine_similarity().

c. Analyze which similarity metric gives better insights in your case.

```python

import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Two texts describing different technologies

text1 = """
Artificial intelligence allows computers to learn from data.
AI helps machines solve complex problems.
It is used in healthcare and education.
Machine learning improves automated decision making.
AI makes technology smarter and faster.
"""

text2 = """
Blockchain is a decentralized digital technology.
It stores information in secure digital blocks.
Blockchain provides transparency and security.
It is widely used in cryptocurrency and finance.
This technology removes the need for central authorities.
"""

# 1. Preprocessing and tokenization
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-z\s]', '', text)
    return text.split()

tokens1 = preprocess(text1)
tokens2 = preprocess(text2)

print("AI Tokens:")
print(tokens1)

print("\nBlockchain Tokens:")
print(tokens2)

# 2(a). Jaccard Similarity
set1 = set(tokens1)
set2 = set(tokens2)

intersection = set1.intersection(set2)
union = set1.union(set2)

jaccard = len(intersection) / len(union)

print("\nCommon Words:", intersection)
print("Jaccard Similarity:", jaccard)

# 2(b). Cosine Similarity using TF-IDF
vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform([text1, text2])

cosine = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("Cosine Similarity:", cosine[0][0])

# 2(c). Comparing both metrics
print("\nComparison:")

if cosine[0][0] > jaccard:
    print("Cosine Similarity gives a higher similarity score.")
else:
    print("Jaccard Similarity gives a higher similarity score.")
```

Jaccard Similarity is useful for comparing the common vocabulary between two texts because it considers only the presence or absence of words. Cosine Similarity using TF-IDF is more informative when comparing text documents because it considers the importance of words and their frequency distribution. Therefore, Cosine Similarity provides better insights for comparing the content of these two technology descriptions.

Write a short review for a product or service.
1. Use TextBlob or VADER to find polarity and subjectivity for each review.
2. Classify reviews into Positive / Negative / Neutral.
3. Create a word cloud using the wordcloud library for all positive reviews.

```python

from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import pandas as pd

# Writing product reviews

reviews = [
    "This laptop is amazing and works very fast. I love its performance.",
    "The product is excellent and the battery life is wonderful.",
    "The laptop is terrible and extremely slow. I hate its performance.",
    "The service was disappointing and the product was very poor.",
    "The laptop is okay and has average performance.",
    "The product is neither good nor bad. It is average."
]

# Lists to store results
results = []

positive_reviews = []

# Sentiment analysis
for review in reviews:

    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity
    subjectivity = analysis.sentiment.subjectivity

    # Classifying sentiment
    if polarity > 0:
        sentiment = "Positive"

    elif polarity < 0:
        sentiment = "Negative"

    else:
        sentiment = "Neutral"

    results.append({
        "Review": review,
        "Polarity": round(polarity, 3),
        "Subjectivity": round(subjectivity, 3),
        "Sentiment": sentiment
    })

    if sentiment == "Positive":
        positive_reviews.append(review)


# Displaying sentiment table
sentiment_df = pd.DataFrame(results)

print("SENTIMENT ANALYSIS RESULTS")
display(sentiment_df)


# Creating Word Cloud from positive reviews
positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color="white"
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation="bilinear")
plt.axis("off")
plt.title("Word Cloud of Positive Reviews")
plt.show()
```

Choose your own paragraph (approximately 100 words) as training data.
1. Tokenize text using Tokenizer() from keras.preprocessing.text.
2. Create input sequences and build a simple LSTM or Dense model.
3. Train the model and generate 2–3 new lines of text starting from any seed word you provide.

```python

import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
```

```python

# Training paragraph

training_text = """
Artificial intelligence is changing the world of technology.
It allows computers to learn from information and solve problems.
Machine learning helps systems recognize patterns and make predictions.
Deep learning uses neural networks to process complex information.
Artificial intelligence is used in healthcare, education, business and science.
It helps doctors understand diseases and supports students in learning.
Technology continues to improve through research and innovation.
People use intelligent machines to complete tasks quickly and efficiently.
The future of artificial intelligence offers many opportunities.
Responsible development of technology is important for society.
Computers can support human creativity and decision making.
Artificial intelligence makes the modern world more connected and advanced.
"""

print("Training Text:")
print(training_text)
```

```python

# Creating tokenizer
tokenizer = Tokenizer()

tokenizer.fit_on_texts([training_text])

# Converting text into sequences of numbers
input_sequences = []

for line in training_text.lower().split("."):

    sequence = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(sequence)):
        ngram_sequence = sequence[:i+1]
        input_sequences.append(ngram_sequence)

# Finding maximum sequence length
max_sequence_len = max(
    len(sequence) for sequence in input_sequences
)

# Padding sequences
input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding='pre'
)

# Splitting into input X and output y
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# Total vocabulary size
total_words = len(tokenizer.word_index) + 1

print("Vocabulary Size:", total_words)
print("Maximum Sequence Length:", max_sequence_len)

print("\nInput Sequences:")
print(X[:5])

print("\nOutput Words:")
print(y[:5])
```

```python

# Creating LSTM model
model = Sequential()

# Embedding layer
model.add(
    Embedding(
        input_dim=total_words,
        output_dim=64
    )
)

# LSTM layer
model.add(LSTM(100))

# Output layer
model.add(
    Dense(total_words, activation='softmax')
)

# Compiling the model
model.compile(
    loss='sparse_categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

# Displaying model structure
model.summary()
```

```python

# Training the model
history = model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)

print("\nModel Training Completed!")
```

```python

# Function to generate new text
def generate_text(seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences(
            [seed_text]
        )[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding='pre'
        )

        predicted = model.predict(
            token_list,
            verbose=0
        )

        predicted_word_index = np.argmax(predicted, axis=-1)[0]

        output_word = ""

        for word, index in tokenizer.word_index.items():

            if index == predicted_word_index:
                output_word = word
                break

        if not output_word:
            break

        seed_text += " " + output_word

    return seed_text


# Providing seed words
seed_words = [
    "artificial intelligence",
    "machine learning",
    "technology"
]

# Generating 3 new lines
print("GENERATED TEXT")

for seed in seed_words:

    generated = generate_text(seed, 10)

    print("\n", generated.capitalize() + ".")
```
