Q2: Generate and Score a Hypothesis – Implement a scoring function that takes a query string and returns all matching entries ranked by confidence

```python
def score_query(query, df):
    query_words = query.lower().split()
    results = []

    for _, row in df.iterrows():
        keywords = row["keywords"].lower().split()
        score = sum(word in keywords for word in query_words)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    results = sorted(results, key=lambda x: x["score"], reverse=True)

    return pd.DataFrame(results)
```

```python
score_query("how can i pay fee", faq_df)
```

Q6: Modify the Q2 scoring function so that if two or more entries tie for the highest score, it prints all matching entries instead of silently picking one. Demonstrate with one query that produces a tie and one that doesn't.

```python
def score_query_with_ties(query, df):
    query_words = query.lower().split()
    results = []

    for _, row in df.iterrows():
        keywords = row["keywords"].lower().split()
        score = sum(word in keywords for word in query_words)

        if score > 0:
            results.append({
                "question": row["question"],
                "answer": row["answer"],
                "category": row["category"],
                "score": score
            })

    if not results:
        print("No matching entries found.")
        return

    results_df = pd.DataFrame(results)

    max_score = results_df["score"].max()

    top_matches = results_df[results_df["score"] == max_score]

    print("Highest score:", max_score)
    print("Matching FAQ(s):")
    display(top_matches)
```

```python
score_query_with_ties("fee", faq_df)
```

```python
score_query_with_ties("pay upi", faq_df)
```
