```python
# Q3

import random

random.seed(int(roll_no))

numbers = [random.randint(100, 900) for _ in range(100)]

print("Random numbers:")
print(numbers)
```

```python
odd_numbers = [x for x in numbers if x % 2 != 0]

print("Odd numbers:", odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

```python
even_numbers = [x for x in numbers if x % 2 == 0]

print("Even numbers:", even_numbers)
print("Count of even numbers:", len(even_numbers))
```

```python
def is_prime(n):
    if n < 2:
        return False

    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False

    return True


prime_numbers = [x for x in numbers if is_prime(x)]

print("Prime numbers:", prime_numbers)
print("Count of prime numbers:", len(prime_numbers))
```

```python
frequency = {}

for x in numbers:
    frequency[x] = frequency.get(x, 0) + 1

most_frequent = max(frequency, key=frequency.get)

print("Most frequent number:", most_frequent)
print("Number of times it occurs:", frequency[most_frequent])
```

```python
# Q4

digits = [int(digit) for digit in roll_no[:8]]

A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)
```

```python
union_set = A.union(B)
intersection_set = A.intersection(B)

print("Union of A and B:", union_set)
print("Intersection of A and B:", intersection_set)
```

```python
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference() gives elements present in one set but not the other.
# symmetric_difference() gives elements present in either set but not in both.
```

```python
symmetric_difference = A.symmetric_difference(B)

print("Symmetric difference:", symmetric_difference)
```

```python
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))
```

```python
X = int(input("Enter a value to remove from set A: "))

A.discard(X)

print("Set A after discard:", A)

# discard() does not give an error if the value is not present,
# so it is safer than remove() when we are unsure whether the value exists.
```

```python
# Q5

my_dict = {
    "name": "Sanvi Goel",
    "roll_no": roll_no,
    "branch": "CSE",
    "age": 20,
    "city": "Chandigarh"
}

print(my_dict)
```

```python
my_dict["location"] = my_dict.pop("city")

print(my_dict)
```

```python
my_dict["cgpa"] = 8.5

print(my_dict)
```

```python
my_dict["age"] = my_dict["age"] + 1

print(my_dict)
```

```python
dict_pop = my_dict.copy()

removed_branch = dict_pop.pop("branch")

print("Using pop():")
print(dict_pop)
print("Returned value:", removed_branch)


dict_del = my_dict.copy()

del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop() removes the key and returns its value,
# while del removes the key but does not return its value.
```

```python
for key, value in my_dict.items():
    print(key, "→", value)
```

```python
if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary.")
```

```python
friend_dict = {
    "name": "Aarav",
    "roll_no": "99999999",
    "branch": "CSE",
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}

print("Merged dictionary:")
print(merged_dict)

# When both dictionaries have the same key,
# the value from the second dictionary wins.
```

```python
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}

print("Key-value pairs where value is a string:")
print(string_values)
```
