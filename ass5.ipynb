Write a program to create a NumPy 1-D array with 5 elements and perform basic operations like:

a) Addition of 2 to all the elements.

b) Multiply 3 with all the elements.

c) Divide every element by 2.

```python

import numpy as np

# Creating a 1-D NumPy array with 5 elements
arr = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr)

# a) Addition of 2
print("After adding 2:", arr + 2)

# b) Multiplication by 3
print("After multiplying by 3:", arr * 3)

# c) Division by 2
print("After dividing by 2:", arr / 2)
```

Q2 (a). Reverse the NumPy array
Question: Reverse the NumPy array:

arr = np.array([1, 2, 3, 6, 4, 5])

```python

import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

# Reversing the array
reverse_arr = arr[::-1]

print("Original Array:", arr)
print("Reversed Array:", reverse_arr)
```

Q2 (b). Find the most frequent value and its indices
Question: Find the most frequent value and their indices in the following arrays:

i. x = np.array([1,2,3,4,5,1,2,1,1,1])

```python

import numpy as np

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

# Finding unique values and their frequencies
values, counts = np.unique(x, return_counts=True)

# Finding the maximum frequency
max_count = np.max(counts)

# Finding the most frequent value
most_frequent = values[counts == max_count]

print("Most frequent value:", most_frequent)

# Finding indices
for val in most_frequent:
    print("Indices of", val, ":", np.where(x == val)[0])
```

ii. y = np.array([1,1,1,2,3,4,2,4,3,3])

```python

import numpy as np

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

# Finding unique values and their frequencies
values, counts = np.unique(y, return_counts=True)

# Finding maximum frequency
max_count = np.max(counts)

# Finding most frequent values
most_frequent = values[counts == max_count]

print("Most frequent values:", most_frequent)

# Finding indices of most frequent values
for val in most_frequent:
    print("Indices of", val, ":", np.where(y == val)[0])
```

Q3. Accessing elements in a 2-D NumPy array
Question: For the given 2-D array:

arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

Access elements using row and column indices:

a) Access 1st row, 2nd column.

b) Access 3rd row, 1st column.

```python

import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Original Array:")
print(arr)

# a) 1st row, 2nd column
print("1st row, 2nd column:", arr[0, 1])

# b) 3rd row, 1st column
print("3rd row, 1st column:", arr[2, 0])
```

Q4. Creating a 1-D NumPy array using linspace()
Question: Write a program to create a 1-D NumPy array named your_name with evenly spaced 25 numbers from 10 to 100 using linspace().

Print:
1. Dimensions of the array.
2. Shape.
3. Total elements.
4. Data type of each element.
5. Total number of bytes consumed by the array.
Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python

import numpy as np

# Creating an array with 25 evenly spaced numbers
your_name = np.linspace(10, 100, 25)

print("Array:", your_name)

# Printing array properties
print("Dimensions:", your_name.ndim)
print("Shape:", your_name.shape)
print("Total elements:", your_name.size)
print("Data type:", your_name.dtype)
print("Total bytes:", your_name.nbytes)

# Reshaping into a 5x5 matrix
reshaped_array = your_name.reshape(5, 5)

print("\nReshaped Array:")
print(reshaped_array)

# Finding transpose
print("\nTranspose using reshape:")
print(reshaped_array.T)
```

Can we use T attribute?

```python

# Transpose using T attribute
print("Transpose using T:")
print(reshaped_array.T)
```

Q5. Creating and reshaping a 2-D NumPy array
Question: Create a 2-D array of three rows and four columns, named ucs420_your_name with the following values:

10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35.

Compute:
- Mean
- Median
- Maximum
- Minimum
- Unique elements
Reshape the array to four rows and three columns and name it reshaped_ucs420_your_name.

Resize the array to two rows and three columns and name it resized_ucs420_your_name.

```python

import numpy as np

# Creating a 2-D array with 3 rows and 4 columns
ucs420_your_name = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_your_name)

# Calculating statistical operations
print("\nMean:", np.mean(ucs420_your_name))
print("Median:", np.median(ucs420_your_name))
print("Maximum:", np.max(ucs420_your_name))
print("Minimum:", np.min(ucs420_your_name))
print("Unique Elements:", np.unique(ucs420_your_name))

# Reshaping into 4 rows and 3 columns
reshaped_ucs420_your_name = ucs420_your_name.reshape(4, 3)

print("\nReshaped Array (4x3):")
print(reshaped_ucs420_your_name)

# Resizing into 2 rows and 3 columns
resized_ucs420_your_name = np.resize(ucs420_your_name, (2, 3))

print("\nResized Array (2x3):")
print(resized_ucs420_your_name)
```
