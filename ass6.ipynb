Question
Consider the following sensor readings:

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

A sensor has an average measurement error of +2°C.

Perform the following without using a for loop:

a. Add 2°C to every reading using vectorization.

b. Convert all temperatures from Celsius to Fahrenheit using:
\[
F = \frac{9}{5}C + 32
\]

c. Identify all readings greater than 32°C using Boolean indexing.

d. Count how many readings exceed 32°C.

e. Explain how vectorization and Boolean indexing can make data processing more efficient than explicitly iterating through every value.



```python

import numpy as np

# Original temperature readings
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

print("Original Temperature:", temperature)

# a) Adding 2°C to every reading
corrected_temperature = temperature + 2

print("Corrected Temperature:", corrected_temperature)

# b) Converting Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32

print("Temperature in Fahrenheit:", fahrenheit)

# c) Identifying readings greater than 32°C
readings_above_32 = corrected_temperature[corrected_temperature > 32]

print("Readings greater than 32°C:", readings_above_32)

# d) Counting readings greater than 32°C
count = np.sum(corrected_temperature > 32)

print("Number of readings above 32°C:", count)
```

Question
The following matrix represents the number of daily steps recorded by four users over three days:
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

Perform the following:

a. Calculate the total steps recorded.

b. Calculate the mean number of steps.

c. Find the maximum and minimum values.

d. Calculate the total steps for each day using axis=0.

e. Calculate the total steps for each user using axis=1.

f. Find the user/day position containing the maximum number of steps using argmax().

```python

import numpy as np

# Creating the steps array
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Steps Matrix:")
print(steps)

# a) Total steps recorded
print("\nTotal Steps:", np.sum(steps))

# b) Mean number of steps
print("Mean Steps:", np.mean(steps))

# c) Maximum and minimum values
print("Maximum Steps:", np.max(steps))
print("Minimum Steps:", np.min(steps))

# d) Total steps for each day
print("Total Steps for Each Day:", np.sum(steps, axis=0))

# e) Total steps for each user
print("Total Steps for Each User:", np.sum(steps, axis=1))

# f) Position of maximum number of steps
position = np.unravel_index(np.argmax(steps), steps.shape)

print("Maximum Steps Position:", position)
```

Question
Write a Python program using NumPy to perform the following operations:

a. Create the following NumPy array:

original = np.array([1, 2, 3, 4, 5, 6])

b. Create a slice containing elements from index 1 to 4 and store it in a variable named subset.

c. Modify the first element of subset to 999 and display both original and subset.

d. Create another slice from original, but this time use .copy(). Modify its first element to 500 and display original and copied array.

e. Create a NumPy array containing numbers from 1 to 12 using np.arange() and reshape it into a 3 × 4 matrix.

f. From the 3 × 4 matrix, extract the following using NumPy indexing and slicing:
- First row
- Last row
- Second column
- Elements from rows 1–2 and columns 2–3
g. Create a flattened version of the 3 × 4 matrix using both flatten() and ravel().

h. Modify an element in the array returned by ravel() and observe whether the original array is affected.

i. Modify an element in the array returned by flatten() and observe whether the original array is affected.

j. Display the shape, ndim, size, and dtype of the original 3 × 4 matrix.

```python

import numpy as np

# a) Creating original array
original = np.array([1, 2, 3, 4, 5, 6])

print("Original Array:", original)

# b) Creating a slice from index 1 to 4
subset = original[1:5]

print("Subset:", subset)

# c) Modifying first element of subset
subset[0] = 999

print("\nAfter modifying subset:")
print("Original Array:", original)
print("Subset:", subset)

# d) Creating another slice using copy()
copied_array = original[1:5].copy()

copied_array[0] = 500

print("\nAfter modifying copied array:")
print("Original Array:", original)
print("Copied Array:", copied_array)

# e) Creating a 3x4 matrix using arange()
matrix = np.arange(1, 13).reshape(3, 4)

print("\n3x4 Matrix:")
print(matrix)

# f) Extracting elements using indexing and slicing

# First row
print("\nFirst Row:", matrix[0, :])

# Last row
print("Last Row:", matrix[-1, :])

# Second column
print("Second Column:", matrix[:, 1])

# Rows 1-2 and columns 2-3
print("Selected Elements:")
print(matrix[1:3, 2:4])

# g) Flattening using flatten() and ravel()
flat_array = matrix.flatten()
ravel_array = matrix.ravel()

print("\nFlatten Array:", flat_array)
print("Ravel Array:", ravel_array)

# h) Modifying ravel array
ravel_array[0] = 100

print("\nAfter modifying ravel:")
print("Original Matrix:")
print(matrix)

# i) Modifying flatten array
flat_array[1] = 200

print("\nAfter modifying flatten:")
print("Flatten Array:", flat_array)
print("Original Matrix:")
print(matrix)

# j) Displaying matrix properties
print("\nShape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Total Elements:", matrix.size)
print("Data Type:", matrix.dtype)
```

Question
A simple cognitive-assistive system wants to predict a user's assistance score based on three factors:
- Sleep hours
- Activity level
- Stress level
Consider:
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])
Here, each row represents one user and each column represents a feature.

Perform the following:

a. Display the shape and dimensions of X.

b. Calculate X.T and explain what the transpose represents.

c. Calculate the matrix product X.T @ X.

d. Calculate the matrix inverse using np.linalg.inv() where appropriate.

e. Implement the Ordinary Least Squares equation:
\[
\beta = (X^T X)^{-1}X^T y
\]

using NumPy.

f. Explain what the resulting coefficients represent in the context of the three user features.

g. A new user has:

new_user = np.array([45, 40, 7])

Use your calculated coefficients to obtain a predicted assistance score.

```python

import numpy as np

# Creating feature matrix X
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

# Creating target array y
y = np.array([40, 65, 30, 85])

# a) Display shape and dimensions
print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

# b) Calculate transpose
X_T = X.T

print("\nTranspose of X:")
print(X_T)

# c) Calculate matrix product X.T @ X
product = X_T @ X

print("\nX.T @ X:")
print(product)

# d) Calculate inverse
inverse = np.linalg.inv(product)

print("\nInverse of X.T @ X:")
print(inverse)

# e) Calculate Ordinary Least Squares coefficients
beta = inverse @ X_T @ y

print("\nCalculated Coefficients (Beta):")
print(beta)

# f) Display the meaning of coefficients
print("\nSleep Coefficient:", beta[0])
print("Activity Coefficient:", beta[1])
print("Stress Coefficient:", beta[2])

# g) Predict assistance score for a new user
new_user = np.array([45, 40, 7])

predicted_score = new_user @ beta

print("\nPredicted Assistance Score:", predicted_score)
```
