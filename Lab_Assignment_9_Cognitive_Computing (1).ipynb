## Question 1: Design a Cognitive Assistant for Reading Labels of Medicine Bottle (You can download Dataset from Internet)

### Aim
To build a vision-based cognitive assistant that reads the text printed on a medicine bottle using Optical Character Recognition (OCR).

### Working
1. Upload an image of a medicine bottle.
2. Use an OCR model to detect and extract text from the image.
3. Display the extracted text so that the user can read the medicine label.

**Note:** This is an educational OCR demonstration, not a medical advice system. Always verify medicine names, dosage, and instructions with the original packaging or a qualified healthcare professional.

```python
# Install the OCR library
!pip -q install easyocr
```

```python
# Import required libraries
import easyocr
import cv2
import matplotlib.pyplot as plt
from google.colab import files
from IPython.display import display, Markdown
import numpy as np
```

```python
# Upload an image of a medicine bottle
uploaded = files.upload()

# Store the uploaded image filename
image_path = next(iter(uploaded.keys()))
print("Uploaded image:", image_path)
```

```python
# Display the uploaded medicine bottle image
image = cv2.imread(image_path)

if image is None:
    raise ValueError("The image could not be opened. Please upload a JPG or PNG image.")

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(7, 7))
plt.imshow(image_rgb)
plt.axis("off")
plt.title("Uploaded Medicine Bottle")
plt.show()
```

```python
# Create an OCR reader and extract text
# EasyOCR supports English text in this experiment.
reader = easyocr.Reader(['en'], gpu=True)
results = reader.readtext(image_path)

print("TEXT DETECTED ON THE MEDICINE LABEL:\n")

if len(results) == 0:
    print("No text was detected. Try a clearer, well-lit image.")
else:
    for number, result in enumerate(results, start=1):
        detected_text = result[1]
        confidence = result[2]
        print(f"{number}. {detected_text}  (Confidence: {confidence:.2f})")
```

```python
# Display the extracted text in an easy-to-read format
extracted_lines = [result[1] for result in results]
extracted_text = "\\n".join(extracted_lines)

if extracted_text:
    display(Markdown("### Medicine Label Text\\n" + extracted_text.replace("\\n", "  \\n")))
else:
    display(Markdown("**No readable text found. Please upload a clearer image.**"))
```

### Result – Question 1
The system accepts an image of a medicine bottle and uses Optical Character Recognition (OCR) to identify and extract the printed text. The detected text is displayed to the user.

### Conclusion – Question 1
Thus, a basic vision-based cognitive assistant for reading medicine bottle labels was implemented using Python and EasyOCR in Google Colab.

## Question 2: Design an Emotion-aware Cognitive Assistant using Facial Expression Detection

### Aim
To build an emotion-aware cognitive assistant that analyses a face in an uploaded image and predicts the facial expression.

### Working
1. Upload an image containing a clearly visible face.
2. Detect the face in the image.
3. Analyse the facial expression using a pre-trained deep-learning model.
4. Display the predicted emotion and its confidence score.

**Note:** Facial-expression models make estimates from visible patterns. They cannot reliably determine a person's actual feelings, personality, or mental state.

```python
# Install DeepFace and its compatibility package
!pip -q install deepface tf-keras
```

```python
# Import required libraries
from deepface import DeepFace
from google.colab import files
from IPython.display import display
import cv2
import matplotlib.pyplot as plt
```

```python
# Upload an image containing a face
uploaded_face = files.upload()

face_path = next(iter(uploaded_face.keys()))
print("Uploaded image:", face_path)
```

```python
# Display the uploaded face image
face_image = cv2.imread(face_path)

if face_image is None:
    raise ValueError("The image could not be opened. Please upload a JPG or PNG image.")

face_rgb = cv2.cvtColor(face_image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(face_rgb)
plt.axis("off")
plt.title("Uploaded Face Image")
plt.show()
```

```python
# Analyse the facial expression
# The model returns estimated scores for emotion categories.
analysis = DeepFace.analyze(
    img_path=face_path,
    actions=['emotion'],
    detector_backend='opencv',
    enforce_detection=False
)

# DeepFace may return one result or a list of results.
if isinstance(analysis, list):
    face_result = analysis[0]
else:
    face_result = analysis

emotion_scores = face_result['emotion']
dominant_emotion = face_result['dominant_emotion']

print("EMOTION ANALYSIS RESULT")
print("-----------------------")
print("Predicted facial expression:", dominant_emotion)

print("\\nEmotion confidence scores:")
for emotion, score in sorted(emotion_scores.items(), key=lambda item: item[1], reverse=True):
    print(f"{emotion}: {score:.2f}%")
```

```python
# Display the predicted emotion as a simple assistant response
assistant_messages = {
    "happy": "The facial-expression model detects a happy expression.",
    "sad": "The facial-expression model detects a sad expression.",
    "angry": "The facial-expression model detects an angry expression.",
    "surprise": "The facial-expression model detects a surprised expression.",
    "fear": "The facial-expression model detects a fearful expression.",
    "disgust": "The facial-expression model detects a disgusted expression.",
    "neutral": "The facial-expression model detects a neutral expression."
}

message = assistant_messages.get(
    dominant_emotion.lower(),
    f"The model detects the expression: {dominant_emotion}."
)

display(Markdown("### Cognitive Assistant Response\\n" + message))
print("\\nRemember: this is only an automated estimate of facial expression, not a reliable reading of someone's feelings.")
```

### Result – Question 2
The system accepts a face image and uses a pre-trained facial-expression analysis model to estimate the dominant expression. It displays the predicted expression along with confidence scores for the available emotion categories.

### Conclusion – Question 2
Thus, an emotion-aware cognitive assistant using facial expression detection was implemented using Python and DeepFace in Google Colab. The output represents an estimate from the image and should not be treated as a definite statement about a person's feelings.
